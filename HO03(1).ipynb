```python
%pip install -qU "pyterrier[java]" ir-datasets ir-measures pandas

import pyterrier as pt
import ir_datasets
import pandas as pd
```

```python
dataset = ir_datasets.load("cord19/trec-covid/round1")
```

```python
# Consultas: somente o título de cada tópico
topics = pd.DataFrame([
    {
        "qid": str(q.query_id),
        "query": q.title
    }
    for q in dataset.queries_iter()
])

print("Quantidade de consultas:", len(topics))
display(topics.head())
```

```python
# Julgamentos de relevância
qrels = pd.DataFrame([
    {
        "qid": str(q.query_id),
        "docno": q.doc_id,
        "label": q.relevance
    }
    for q in dataset.qrels_iter()
])

print("Quantidade de julgamentos:", len(qrels))
display(qrels.head())
```

```python
def documentos():
    for doc in dataset.docs_iter():
        yield {
            "docno": str(doc.doc_id),
            "title": doc.title or "",
            "abstract": doc.abstract or ""
        }

indexer = pt.IterDictIndexer(
    "./indice_trec_covid_round1",
    overwrite=True,
    text_attrs=["title", "abstract"]
)

index_ref = indexer.index(documentos())
```

```python
modelo_vetorial = pt.terrier.Retriever(
    index_ref,
    wmodel="TF_IDF",
    num_results=51078
)

modelo_bm25 = pt.terrier.Retriever(
    index_ref,
    wmodel="BM25",
    num_results=51078
)
```

```python
metricas = [
    "map",
    "map_cut.1",
    "map_cut.3",
    "map_cut.5",
    "map_cut.10",
    "ndcg",
    "ndcg_cut.1",
    "ndcg_cut.3",
    "ndcg_cut.5",
    "ndcg_cut.10"
]

resultados = pt.Experiment(
    [modelo_vetorial, modelo_bm25],
    topics,
    qrels,
    eval_metrics=metricas,
    names=["Vetorial TF-IDF", "BM25"],
    verbose=True
)

display(resultados.round(4))
```

```python
tabela = resultados.set_index("name").T

tabela.columns.name = "Modelo"

display(tabela.round(4))

resultados.to_csv(
    "resultados_trec_covid_round1.csv",
    index=False
)

tabela.round(4).to_csv(
    "tabela_comparativa.csv"
)
```

## Conclusão

Com base nos resultados obtidos, o BM25 apresentou um desempenho ligeiramente superior ao modelo vetorial TF-IDF na maioria das métricas avaliadas. O BM25 obteve MAP de 0,1817 e nDCG de 0,5290, enquanto o TF-IDF alcançou 0,1804 e 0,5279, respectivamente.

Nas métricas com cortes em 1, 3, 5 e 10 posições, os dois modelos apresentaram resultados próximos. O BM25 teve valores um pouco maiores em MAP@3, MAP@5, MAP@10, nDCG@5 e nDCG@10. Já o TF-IDF apresentou um resultado melhor em nDCG@3. Na primeira posição, os dois modelos tiveram o mesmo desempenho.

Portanto, o BM25 apresentou uma pequena vantagem geral, mas a diferença entre os modelos foi reduzida. Isso mostra que ambos tiveram resultados semelhantes, embora o BM25 tenha se destacado na maioria das métricas. Assim, não é possível afirmar que um modelo seja melhor em todas as situações, pois o desempenho depende também da métrica utilizada.
